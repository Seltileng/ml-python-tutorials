```python
# Import core data processing and modeling libraries
import pandas as pd 
import numpy as np 
from sklearn.model_selection import train_test_split
```

```python
# Load the 500-hits baseball player dataset with Latin-1 encoding
df = pd.read_csv('500hits.csv', encoding='latin-1')

# Preview the first 10 player records
df.head(10)
```

```python
# Check the dimensions of the dataset (rows, columns)
df.shape
```

```python
# Verify if any feature columns contain missing values
df.isnull().sum()
```

```python
# Feature Selection: Select career baseball batting performance metrics
# YRS: Years active, G: Games, AB: At Bats, R: Runs, H: Hits, 2B: Doubles, 
# 3B: Triples, HR: Home Runs, RBI: Runs Batted In, BB: Base on Balls (Walks), 
# SO: Strikeouts, SB: Stolen Bases, CS: Caught Stealing, BA: Batting Average
x = df[
    [
        'YRS',
        'G',
        'AB',
        'R',
        'H',
        '2B',
        '3B',
        'HR',
        'RBI',
        'BB',
        'SO',
        'SB',
        'CS',
        'BA',
    ]
]

# Target variable: Hall of Fame induction status (1 = Inducted, 0 = Not Inducted)
y = df['HOF']
```

```python
# Examine target class distribution (HOF: 1 vs 0) to evaluate class balance
print(df['HOF'].value_counts())
```

```python
# Split dataset into training (80%) and testing (20%) partitions with random_state=42
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# Instantiate the Logistic Regression classifier with increased iterations for convergence
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=1000)
```

```python
# Train the Logistic Regression model using the training feature set and labels
model.fit(X_train, y_train)
```

```python
# Predict binary classification labels (HOF: 0 or 1) for test set players
y_pred = model.predict(X_test)  
print("Predicted classes:\n", y_pred)
```

```python
# Construct DataFrame comparing actual player HOF status with model predictions
comparison_df = pd.DataFrame({'Actual': y_test, 'Predicted': y_pred})   
print(comparison_df.head(15))
```

```python
# Measure overall model accuracy (fraction of total correct predictions)
from sklearn.metrics import accuracy_score
acc = accuracy_score(y_test, y_pred)
print("Accuracy:", acc)  
print(f"Accuracy Percentage: {acc * 100:.2f}%")
```

```python
# Compute Confusion Matrix to evaluate True Positives, False Positives, True Negatives, False Negatives
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:\n", cm)
```

```python
# Compute Precision: proportion of predicted HOF inductees that were actually inducted (TP / (TP + FP))
from sklearn.metrics import precision_score
prec = precision_score(y_test, y_pred)
print("Precision:", prec)
print(f"Precision Percentage: {prec * 100:.2f}%")
```

```python
# Compute Recall: proportion of actual HOF inductees correctly identified by model (TP / (TP + FN))
from sklearn.metrics import recall_score
rec = recall_score(y_test, y_pred) 
print("Recall:", rec)
print(f"Recall Percentage: {rec * 100:.2f}%")
```

```python
# Compute F1-Score: harmonic mean of precision and recall
from sklearn.metrics import f1_score
f1 = f1_score(y_test, y_pred)
print("F1 Score:", f1)
print(f"F1 Score Percentage: {f1 * 100:.2f}%")
```

```python
# Model Interpretation: Extract model intercept, coefficients, and compute Odds Ratios
intercept = model.intercept_[0]  # Baseline log-odds constant
coefficients = model.coef_[0]    # Log-odds weights for each baseball feature
features = x.columns             # Feature names

# Compute odds ratios: exp(coef) indicates multiplicative change in odds of HOF per 1-unit feature increase
odds_ratios = np.exp(coefficients)

print(f"Model Intercept: {intercept:.4f}\n")
print(f"{'Feature':<10} | {'Coefficient':>12} | {'Odds Ratio':>12}")
print("-" * 40)
for feature, coef, odds in zip(features, coefficients, odds_ratios):
    print(f"{feature:<10} | {coef:>12.4f} | {odds:>12.4f}")
```
