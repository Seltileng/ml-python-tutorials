```python
# Import core libraries for tabular manipulation, modeling, and plotting
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error
```

```python
# Load the house pricing dataset
data = pd.read_csv("House_Pricing_Regression_300_Rows-1.csv")

# Preview the initial records
data.head(301)
```

```python
# Display summary statistics (count, mean, std, min, percentiles, max)
data.describe()
```

```python
# Inspect DataFrame summary: data types and non-null values
data.info()
```

```python
# Check for missing (NaN) values in the dataset
data.isnull().sum()
```

```python
# Separate feature matrix (x) and target price vector (y)
# Exclude House_ID (non-predictive index) and Price_NGN (target label)
x = data.drop(columns=["House_ID", "Price_NGN"])
y = data['Price_NGN']
```

```python
# Partition the dataset into training (80%) and testing (20%) sets
from sklearn.model_selection import train_test_split
X_Train, X_Test, Y_Train, Y_Test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# Initialize and train the Linear Regression model
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_Train, Y_Train)
```

```python
# Generate house price predictions on both test and train sets
y_pred = model.predict(X_Test)
x_pred = model.predict(X_Train)
```

```python
# Evaluate model accuracy using the R-squared (R2) metric
score = r2_score(Y_Test, y_pred)
print(f"Model R-squared Score: {score:.4f}")
```

```python
# Visualize model predictions: Scatter plot of Actual vs. Predicted Prices
plt.figure(figsize=(10, 6))
plt.scatter(Y_Test, y_pred, color='blue', alpha=0.6, label='Predictions')
plt.grid(True)
plt.xlabel("Actual Prices (NGN)")
plt.ylabel("Predicted Prices (NGN)")
plt.title("Actual vs Predicted Prices")

# Draw 45-degree diagonal reference line representing perfect prediction (y = x)
plt.plot([Y_Test.min(), Y_Test.max()], [Y_Test.min(), Y_Test.max()], lw=3, color='red', label='Ideal Fit')
plt.legend()
plt.show()
```
